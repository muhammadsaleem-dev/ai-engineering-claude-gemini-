# Temperature: Controlling Output Randomness

Temperature is a decimal value between `0.0` and `1.0` that influences the **probability distribution** when sampling the next token.

### How Temperature Works:
- **Low Temperature (near 0.0)**:
  - Sharpens the probability distribution toward the #1 most likely token.
  - At `0.0`, sampling is **deterministic** (greedy): always picks the highest-probability token.
  - Best for: Factual responses, data extraction, code generation, moderation.
- **High Temperature (near 1.0)**:
  - Flattens the probability distribution across all possible tokens.
  - Lower-probability tokens have a higher chance of being chosen, producing **creative & varied** output.
  - Best for: Brainstorming, creative writing, marketing copy, joke generation.

```python
# Load environment variables
from dotenv import load_dotenv
load_dotenv()
```

```python
# Initialize Anthropic Client
from anthropic import Anthropic

client = Anthropic()
model = "claude-3-5-sonnet-latest"
```

```python
# Helper functions with dynamic temperature support (default = 1.0)
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None, temperature=1.0):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature,
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
```

## 1. Low Temperature (`temperature=0.0`)
Output is deterministic and reproducible. Repeating this prompt will return similar or identical concepts.

```python
messages = []
add_user_message(messages, "Generate a one sentence movie idea")
answer = chat(messages, temperature=0.0)
print(answer)
```

## 2. High Temperature (`temperature=1.0`)
Output is more diverse, creative, and exploratory across repeated runs.

```python
messages = []
add_user_message(messages, "Generate a one sentence movie idea")
answer = chat(messages, temperature=1.0)
print(answer)
```
