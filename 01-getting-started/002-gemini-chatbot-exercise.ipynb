# Chat Bot Exercise (Gemini Counterpart)

In this exercise, we build an interactive chatbot running directly inside a Jupyter notebook cell using Python's built-in `input()` function and our Gemini helper functions.

### How It Works:
1. **Prompt**: Prompt the user to enter text using the built-in `input()` function.
2. **Store User Input**: Add the input to our list of `messages` using `add_user_message()`.
3. **Call API**: Send the entire message history to Gemini via `chat(messages)`.
4. **Store Model Response**: Add Gemini's reply to `messages` using `add_model_message()`.
5. **Print Output**: Display the assistant's reply delimited by `---`.
6. **Loop**: Repeat forever until stopped.

```python
# Load environment variables
from dotenv import load_dotenv
load_dotenv()
```

```python
# Initialize Gemini Client
import os
import logging
from google import genai

# Suppress AFC informational notices
logging.getLogger("google_genai.models").setLevel(logging.ERROR)

client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))
model = "gemini-3.5-flash-lite"
```

```python
# Helper Functions for Gemini Messages
def add_user_message(messages, text):
    user_message = {"role": "user", "parts": [{"text": text}]}
    messages.append(user_message)

def add_model_message(messages, text):
    model_message = {"role": "model", "parts": [{"text": text}]}
    messages.append(model_message)

def chat(messages):
    response = client.models.generate_content(
        model=model,
        contents=messages,
        config={"max_output_tokens": 1000},
    )
    return response.text
```

```python
# Make an initial list of messages
messages = []

# Use a 'while True' loop to run the chatbot forever
# (To stop: type 'exit' / 'quit' or click the Interrupt Kernel button in the notebook toolbar)
while True:
    # Get user input
    user_input = input("> ")
    print("> ", user_input)

    # Optional graceful exit condition
    if user_input.strip().lower() in ("exit", "quit"):
        print("Exiting chatbot session.")
        break

    # Guard against empty input (pressing Enter without typing causes a 400 API error)
    if not user_input.strip():
        continue

    # Add user input to the list of messages
    add_user_message(messages, user_input)

    # Call Gemini with the 'chat' function
    answer = chat(messages)

    # Add generated text to the list of messages
    add_model_message(messages, answer)

    # Print the generated text
    print("---")
    print(answer)
    print("---")
```
