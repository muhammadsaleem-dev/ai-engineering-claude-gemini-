# Chat Bot Exercise (Course Notebook: `001_requests_exercise.ipynb`)

In this exercise, we build an interactive chatbot running directly inside a Jupyter notebook cell using Python's built-in `input()` function and our three multi-turn helper functions.

### How It Works:
1. **Prompt**: Prompt the user to enter text using the built-in `input()` function.
2. **Store User Input**: Add the input to our list of `messages` using `add_user_message()`.
3. **Call API**: Send the entire message history to Claude via `chat(messages)`.
4. **Store Assistant Response**: Add Claude's reply to `messages` using `add_assistant_message()`.
5. **Print Output**: Display the assistant's reply delimited by `---`.
6. **Loop**: Repeat forever until stopped.

```python
# Load environment variables
from dotenv import load_dotenv
load_dotenv()
```

```python
# Initialize Anthropic Client
from anthropic import Anthropic

client = Anthropic()
model = "claude-3-5-sonnet-latest"
```

```python
# The Three Course Helper Functions
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
    )
    return message.content[0].text
```

```python
# Make an initial list of messages
messages = []

# Use a 'while True' loop to run the chatbot forever
# (To stop: type 'exit' / 'quit' or click the Interrupt Kernel button in the notebook toolbar)
while True:
    # Get user input
    user_input = input("> ")
    print("> ", user_input)

    # Optional graceful exit condition
    if user_input.strip().lower() in ("exit", "quit"):
        print("Exiting chatbot session.")
        break

    # Guard against empty input (pressing Enter without typing causes a 400 API error)
    if not user_input.strip():
        continue

    # Add user input to the list of messages
    add_user_message(messages, user_input)

    # Call Claude with the 'chat' function
    answer = chat(messages)

    # Add generated text to the list of messages
    add_assistant_message(messages, answer)

    # Print the generated text
    print("---")
    print(answer)
    print("---")
```
