```python
# Create an api client
from anthropic import Anthropic

client = Anthropic()
model = "claude-3-5-sonnet-latest"
```

```python
# Make a Single-Turn Request
# NOTE: In Anthropic, `max_tokens` is MANDATORY. Omitting it raises a TypeError.
message = client.messages.create(
    model=model,
    max_tokens=1000,  # Required by Anthropic: caps maximum tokens in the response
    messages=[
        {
            "role": "user",
            "content": "What is an AI engineer?"
        }
    ]
)
```

```python
# View raw message object
message
```

```python
# Print the generated text response
print(message.content[0].text)
```

## Multi-turn Conversations

### The Stateless Nature of LLM APIs
The Anthropic API and Claude do **not** store any messages or past interactions. Every single API request is completely independent.

If you want to maintain a conversation across multiple turns:
1. You must manually maintain a list of messages (conversation history) in your code.
2. You must provide that entire list of messages with every follow-up request.

Below are three helper functions taught in the course to manage conversation history.

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
    )
    return message.content[0].text
```

```python
# Make a starting list of messages
messages = []

# Add in the initial user question of "Define quantum computing in one sentence"
add_user_message(messages, "Define quantum computing in one sentence")

# Pass the list of messages into 'chat' to get an answer
answer = chat(messages)

# Take the answer and add it as an assistant message into our list
add_assistant_message(messages, answer)

# Add in the user's follow-up question
add_user_message(messages, "Write another sentence")

# Call chat again with the list of messages to get a final answer
answer = chat(messages)
answer
```
