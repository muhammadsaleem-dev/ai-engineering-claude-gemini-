# Temperature: Controlling Output Randomness (Gemini Counterpart)

In Google Gemini, `temperature` is passed inside `config={"temperature": ...}`.

### How Temperature Works:
- **Low Temperature (`0.0`)**:
  - Deterministic greedy sampling.
  - Always selects the highest-probability next token.
  - Best for: Data extraction, coding, fact retrieval, structured JSON.
- **High Temperature (`1.0`)**:
  - Flattens token probabilities to encourage diversity.
  - Best for: Brainstorming, creative writing, marketing ideas.

```python
# Load environment variables
from dotenv import load_dotenv
load_dotenv()
```

```python
# Initialize Gemini Client
import os
import logging
from google import genai

logging.getLogger("google_genai.models").setLevel(logging.ERROR)

client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))
model = "gemini-3.5-flash-lite"
```

```python
# Helper functions with dynamic temperature support (default = 1.0)
def add_user_message(messages, text):
    user_message = {"role": "user", "parts": [{"text": text}]}
    messages.append(user_message)

def add_model_message(messages, text):
    model_message = {"role": "model", "parts": [{"text": text}]}
    messages.append(model_message)

def chat(messages, system=None, temperature=1.0):
    config = {
        "max_output_tokens": 1000,
        "temperature": temperature,
    }

    if system:
        config["system_instruction"] = system

    response = client.models.generate_content(
        model=model,
        contents=messages,
        config=config,
    )
    return response.text
```

## 1. Low Temperature (`temperature=0.0`)
Notice the consistent, deterministic themes when running with temperature 0.0.

```python
messages = []
add_user_message(messages, "Generate a one sentence movie idea")
answer = chat(messages, temperature=0.0)
print(answer.strip())
```

## 2. High Temperature (`temperature=1.0`)
Notice the higher creativity, varied vocabulary, and unexpected premises.

```python
messages = []
add_user_message(messages, "Generate a one sentence movie idea")
answer = chat(messages, temperature=1.0)
print(answer.strip())
```
